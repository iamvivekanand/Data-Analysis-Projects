# Data Analysis Fundamentals

## 1. Types of Data

* **Quantitative Data (Numerical):** Data that can be measured or counted objectively.
  * **Discrete Data:** Whole numbers (e.g., number of store visitors, count of items).
  * **Continuous Data:** Values measured on a continuous scale (e.g., temperature, age, height).
* **Qualitative Data (Categorical):** Descriptive, non-numerical data explaining qualities, attributes, or opinions (e.g., customer feedback: *"satisfied"*, *"dissatisfied"*).
* **Structured Data:** Highly organized in predefined formats (rows & columns). Easily queried using SQL or managed in spreadsheets (e.g., bank transactions, employee records).
* **Unstructured Data:** Raw information without a predefined structure. Accounts for over **80% of global data** and requires AI/NLP tools for processing (e.g., emails, images, videos, audio).
* **Semi-Structured Data:** Lacks a rigid relational structure but uses tags/markers for organization (e.g., JSON files, XML documents, NoSQL databases).

---

## 2. Big Data & The 5 Vs

Big Data refers to massive, complex datasets that surpass the capabilities of traditional database management tools. It is defined by the **5 Vs**:

1. **Volume:** Extreme amounts of generated and stored data.
2. **Velocity:** The rapid speed at which data is created and processed.
3. **Variety:** Diverse formats ranging from structured tables to unstructured media.
4. **Veracity:** The accuracy, quality, and trustworthiness of the data.
5. **Value:** The actionable insights extracted from the data to drive decision-making.

---

## 3. Data Lifecycle & Infrastructure

* **Data Collection:** Gathering raw inputs from multiple sources (e.g., website clicks, survey responses, IoT sensors).
* **Data Management:** Standardizing, cleaning, and maintaining data quality and consistency across teams.
* **Data Security:** Protecting sensitive information from unauthorized access or misuse using tools like **encryption**, authentication, and access policies.

---

## 4. Role of Data in AI & Machine Learning

* **Core Foundation:** Machine Learning models are directly dependent on the quality and relevance of their training data.
* **Garbage In, Garbage Out:** Poor, incomplete, or biased data leads to inaccurate or unfair outcomes.
* **Real-world Examples:** 
  * Recommendation engines (e.g., Netflix watch history).
  * Autonomous vehicles navigating via multi-sensor data streams.

---

## 5. Key Applications across Domains

* **Business Analytics:** Trend identification, operational optimization, and strategic decision-making.
* **Healthcare:** Disease prediction, automated diagnostics, and personalized treatment plans.
* **Finance:** Real-time fraud detection, algorithmic trading, and credit scoring.
* **Marketing:** Hyper-targeted campaigns, customer segment behavior analysis, and personalization.

---

## 6. Major Challenges in Data Analysis

* **Data Quality & Accuracy:** Dirty or missing data skews insights and results in bad decision-making.
* **Privacy & Ethics:** Managing sensitive user data while staying compliant with legal frameworks like GDPR and CCPA.
* **Algorithmic Bias:** Historical bias present in training data propagates unfair decisions in ML models.
* **Scalability & Cost:** Managing high-volume storage and complex processing infrastructure is resource-intensive.
