### Outlier Detection & Treatment

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

tips.head()
```

```python
# First: Understand the Distribution
tips["total_bill"].describe()
```

```python
# Calculate Q1 and Q3
# For total_bill:

Q1 = tips["total_bill"].quantile(0.25)
Q3 = tips["total_bill"].quantile(0.75)

Q1, Q3
```

```python
# We can also calculate them together:
tips["total_bill"].quantile([0.25, 0.50, 0.75])
```

```python
# IQR — Interquartile Range
# The most common outlier detection method we will learn first is the IQR method.

IQR = Q3 - Q1

IQR
```

```python
# 9. The IQR Outlier Rule
```

```python
# 10. Calculate the Bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

lower_bound, upper_bound
```

```python
# 11. Find the Outliers
outliers = tips[
    (tips["total_bill"] < lower_bound) |
    (tips["total_bill"] > upper_bound)]

outliers
```
