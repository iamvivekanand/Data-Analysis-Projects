```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

tips.head()
```

```python
# Missing Values Detect करना
tips.isnull() # यह हर cell के लिए True False देगा। लेकिन यह पूरा DataFrame बहुत बड़ा हो सकता है।
```

```python
tips.isnull().sum()
# यह हर column में missing values की count देगा।
```

```python
# 7. Missing Percentage
```

```python
missing_pct = tips.isnull().mean() * 100
missing_pct
```

```python
# Complete Missing Value Report
missing_report = pd.DataFrame({
    "missing_count": tips.isnull().sum(),
    "missing_percentage": tips.isnull().mean() * 100})

missing_report
```

```python
# Practice Dataset बनाते हैं
df = pd.DataFrame({
    "Employee": ["A", "B", "C", "D", "E"],
    "Age": [25, 30, np.nan, 40, 35],
    "Salary": [30000, np.nan, 40000, 50000, np.nan],
    "Department": ["QA", "Production", "QA", "Production", "QA"]})

df
```

```python
df.isnull().sum()
```

```python
df.isnull().mean() * 100
```

```python
# मान लो हमें केवल वे rows चाहिए जिनमें Age missing है।
df[df["Age"].isnull()]
```

```python
# Non-Missing Rows
df[df["Age"].notnull()]
```

```python
df[df["Age"].notna()]
```

```python
# अगर किसी row में किसी भी column में missing value है
df[df.isnull().any(axis=1)]
```

```python
df[df.isnull().all(axis=1)]
```

```python
df.dropna()
```

```python
clean_df = df.dropna()
```

```python
# Specific Column पर dropna()
# अगर हमें सिर्फ Age missing rows हटानी हैं:
df.dropna(subset=["Age"])
```

```python
# Multiple Columns पर dropna()
df.dropna(subset=["Age", "Salary"])
```

```python
# अगर row के सभी values missing हों तभी remove करना हो:
df.dropna(how="all")
```

```python
# Mean कब Use करें?
```

```python
# Median Imputation

median_age = df["Age"].median()
df["Age"] = df["Age"].fillna(median_age)
```

```python
# Mode Imputation
```

```python
# Forward Fill — ffill()
temperature = pd.Series([25, 26, np.nan, np.nan, 29])
temperature
```

```python
temperature.ffill()
# fills with previous value in forward direction 
```

```python
temperature.bfill()
```

```python
# Duplicate Data
df.duplicated()
```

```python
df.duplicated().sum()
```

```python
# View duplicate rows
df[df.duplicated()]
```

```python
# remove duplicates
df.drop_duplicates()
```

```python
clean_df = df.drop_duplicates()
```

```python
# Duplicate Based on Specific Columns
# मान लो employee name unique होना चाहिए

df.duplicated(subset=["Employee"])
```

```python
df.drop_duplicates(
    subset=["Employee"])
```

```python
# Data Type Problems
df.dtypes
```

```python
# in real world data, numerical values may come in form of string
df["Salary"].dtype
```

```python
# 38. String को Numeric में Convert करना
df["Salary"] = pd.to_numeric(
    df["Salary"],
    errors="coerce")
```

errors="coerce" invalid values को NaN बना देगा।
