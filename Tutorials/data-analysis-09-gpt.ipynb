```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

tips.head()
```

```python
tips.info()
```

```python
# 3. What is Data Transformation?
```

```python
# Creating Calculated Columns
tips["tip_percentage"] = tips["tip"] / tips["total_bill"] * 100
```

```python
tips[["total_bill", "tip", "tip_percentage"]].head()
```

```python
tips["tip_percentage"] = tips["tip_percentage"].round(2)
```

```python
tips[["total_bill", "tip", "tip_percentage"]].head()
```

```python
#Bill Per Person
tips["bill_per_person"] = tips["total_bill"] / tips["size"]
```

```python
tips["bill_per_person"] = tips["bill_per_person"].round(2)
tips[["total_bill", "size", "bill_per_person"]].head()
```

```python
# 6. apply() — Transforming Values
def bill_category(bill):
    if bill > 30:
        return "High"
    else:
        return "Normal"
```

```python
tips["bill_category"] = tips["total_bill"].apply(bill_category)
```

```python
tips[["total_bill", "bill_category"]].head(10)
```

```python
# 7. apply() with Lambda
tips["bill_category"] = tips["total_bill"].apply(
    lambda x: "High" if x > 30 else "Normal")
```

```python
# apply() with Multiple Conditions
def classify_bill(x):
    if x < 15:
        return "Low"
    elif x <= 30:
        return "Medium"
    else:
        return "High"
```

```python
tips["bill_category"] = tips["total_bill"].apply(classify_bill)
```

```python
tips[["total_bill", "bill_category"]].head(5)
```

```python
# 9. apply() with axis=1
def tip_level(row):
    if row["tip_percentage"] >= 20:
        return "Excellent"
    elif row["tip_percentage"] >= 15:
        return "Good"
    else:
        return "Low"

tips["tip_level"] = tips.apply(tip_level, axis=1)
```

```python
# 10. map() — Mapping Values
```

```python
tips["sex_code"] = tips["sex"].map({
    "Male": 1,
    "Female": 0})
```

```python
tips[["sex", "sex_code"]].head()
```

```python
# 11. map() vs apply()
```

```python
tips["sex_code"] = tips["sex"].map({"Male": 1,"Female": 0})
```

```python
# apply(), is More flexible.
tips["bill_category"] = tips["total_bill"].apply(
    lambda x: "High" if x > 30 else "Normal")
```
