#### Data Analysis Bootcamp — Lesson 4
#### Data Filtering & Conditional Analysis with Pandas

```python
tips["total_bill"] > 20
```

```python
tips[tips["total_bill"] > 20][0:4]
```

```python
tips[tips["sex"] == "Male"][0:4]
```

```python
tips[tips["sex"] != "Male"][0:4]
```

```python
tips[tips["smoker"] == "Yes"][0:4]
```

```python
# Show all customers who visited on Sunday.
tips[tips["day"] == "Sun"][0:3]
```

```python
# How many customers visited on Sunday?
len(tips[tips["day"] == "Sun"])
```

```python
(tips["day"] == "Sun").sum()
```

#### Multiple Conditions

```python
# Customers whose bill is greater than $20 AND tip is greater than $3.
tips[(tips["total_bill"] > 20) & (tips["tip"] > 3)][0:4]
```

```python
tips[(tips["sex"] == "Male") & (tips["smoker"] == "Yes")][0:4]
```

```python
tips[
    (tips["sex"] == "Male") |
    (tips["smoker"] == "Yes")
]
```

```python
# Multiple AND Conditions
tips[
    (tips["total_bill"] > 20) &
    (tips["tip"] > 3) &
    (tips["size"] >= 3)
]
```

```python
# Suppose we want customers who came on:Saturday OR Sunday
tips[
    (tips["day"] == "Sat") |
    (tips["day"] == "Sun")
]
```

```python
#  or
tips[tips["day"].isin(["Sat", "Sun"])]
```

```python
tips[tips["day"].isin(["Fri", "Sat", "Sun"])]
```

```python
# Day is NOT Friday or Saturday.
tips[~tips["day"].isin(["Fri", "Sat"])]
```

```python
tips.loc[
    tips["total_bill"] > 20,
    ["total_bill", "tip", "day"]
]
```

```python
# Show bills greater than $20, highest bill first.
tips[tips["total_bill"] > 20].sort_values("total_bill",ascending=False)
```
